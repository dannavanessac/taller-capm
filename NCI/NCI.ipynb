```python
import pandas as pd
df_nci = pd.read_csv('/content/drive/MyDrive/2026 - 2/DEFI II/TALLER BTC/NCI.csv')
display(df_nci.head())
```

### Añadir nuevas columnas al dataset

```python
import numpy as np
```

```python
# 1. Retornos Logarítmicos
df_nci['Log_Returns'] = np.log(df_nci['Close/Last'] / df_nci['Close/Last'].shift(1))

# Mostrar las primeras filas con las nuevas columnas
print("DataFrame con las nuevas columnas:")
display(df_nci.head())
```

### Crear una tabla aparte con estadísticas

```python
# Calcular las métricas solicitadas

# Retorno promedio diario
mean_daily_return = df_nci['Log_Returns'].mean()

# Retorno acumulado
cumulative_return = (df_nci['Close/Last'].iloc[-1] / df_nci['Close/Last'].iloc[0]) - 1

# Retorno anualizado (asumiendo 252 días hábiles en un año para retornos logarítmicos)
annualized_return = mean_daily_return * 252

# Desviación estándar de retornos diarios
std_daily_returns = df_nci['Log_Returns'].std()

# Volatilidad anualizada
annualized_volatility = std_daily_returns * np.sqrt(252)

# Varianza de retornos diarios
variance_daily_returns = df_nci['Log_Returns'].var()

# Asimetría de retornos diarios
skewness_daily_returns = df_nci['Log_Returns'].skew()

# Curtosis de retornos diarios
kurtosis_daily_returns = df_nci['Log_Returns'].kurt()

# Crear un DataFrame para las estadísticas
stats_df = pd.DataFrame({
    'Métrica': [
        'Retorno promedio diario',
        'Retorno acumulado',
        'Retorno anualizado',
        'Desviación estándar de retornos diarios',
        'Volatilidad anualizada',
        'Varianza de retornos diarios',
        'Asimetría de retornos diarios',
        'Curtosis de retornos diarios'
    ],
    'Valor': [
        mean_daily_return,
        cumulative_return,
        annualized_return,
        std_daily_returns,
        annualized_volatility,
        variance_daily_returns,
        skewness_daily_returns,
        kurtosis_daily_returns
    ]
})

# Mostrar la tabla de estadísticas
print("Tabla de Estadísticas de Retornos de XBI:")
display(stats_df)
```

```python
df_nci.to_csv('df_nci.csv', index=True)
stats_df.to_csv('nci_stats.csv', index=False)
print("df_xbi.csv and xbi_stats.csv have been downloaded.")
```
