### Descargar datos de Gossamer Bio

```python
# Instalar la librería yfinance para descargar datos financieros
!pip install yfinance
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

# Definir el ticker de Gossamer
ticker = 'GOSS'

# Definir la fecha de corte (hasta el 24 de agosto de 2026)
end_date = '2026-08-24'

# Definir la fecha de inicio (desde el 25 de agosto de 2021)
start_date = '2021-08-25'

# Descargar los datos históricos
df_goss = yf.download(ticker, start=start_date, end=end_date)

# Mostrar las primeras filas del DataFrame
print(f"\n\nDatos de Gossamer Bio Inc. desde {start_date} hasta {end_date}:")
display(df_goss.head())
```

### Añadir nuevas columnas al dataset

```python
# 1. Retornos Logarítmicos
df_goss['Log_Returns'] = np.log(df_goss['Close'] / df_goss['Close'].shift(1))

# 2. Volumen Monetario
df_goss['Monetary_Volume'] = df_goss['Volume'] * df_goss['Close']

# 3. Bid-Ask Spread
df_goss['Bid_Ask_Spread'] = df_goss['High'] - df_goss['Low']

# 4. Capitalización de Mercado (aproximación)
# Nota: La verdadera capitalización de mercado requiere el 'circulating supply' del activo.
# Aquí se calcula como el 'Close' multiplicado por el 'Volume' (valor monetario transado).
df_goss['Capitalizacion_Mercado'] = df_goss['Close'] * df_goss['Volume']

# Mostrar las primeras filas con las nuevas columnas
print("DataFrame con las nuevas columnas:")
display(df_goss.head())
```

### Crear una tabla aparte con estadísticas

```python
# Calcular las métricas solicitadas

# Retorno promedio diario
mean_daily_return = df_goss['Log_Returns'].mean()

# Retorno acumulado
cumulative_return = (df_goss['Close'].iloc[-1] / df_goss['Close'].iloc[0]) - 1

# Retorno anualizado (asumiendo 252 días hábiles en un año para retornos logarítmicos)
annualized_return = mean_daily_return * 252

# Desviación estándar de retornos diarios
std_daily_returns = df_goss['Log_Returns'].std()

# Volatilidad anualizada
annualized_volatility = std_daily_returns * np.sqrt(252)

# Varianza de retornos diarios
variance_daily_returns = df_goss['Log_Returns'].var()

# Asimetría de retornos diarios
skewness_daily_returns = df_goss['Log_Returns'].skew()

# Curtosis de retornos diarios
kurtosis_daily_returns = df_goss['Log_Returns'].kurt()

# Crear un DataFrame para las estadísticas
stats_df = pd.DataFrame({
    'Métrica': [
        'Retorno promedio diario',
        'Retorno acumulado',
        'Retorno anualizado',
        'Desviación estándar de retornos diarios',
        'Volatilidad anualizada',
        'Varianza de retornos diarios',
        'Asimetría de retornos diarios',
        'Curtosis de retornos diarios'
    ],
    'Valor': [
        mean_daily_return,
        cumulative_return,
        annualized_return,
        std_daily_returns,
        annualized_volatility,
        variance_daily_returns,
        skewness_daily_returns,
        kurtosis_daily_returns
    ]
})

# Mostrar la tabla de estadísticas
print("Tabla de Estadísticas de Retornos de Gossamer Bio Inc.:")
display(stats_df)
```

```python
df_goss.to_csv('df_goss.csv', index=True)
stats_df.to_csv('goss_stats.csv', index=False)
print("df_xbi.csv and xbi_stats.csv have been downloaded.")
```
