### Descargar datos de NBI

```python
# Instalar la librería yfinance para descargar datos financieros
!pip install yfinance
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

# Definir el ticker de NBI
ticker = '^NBI'

# Definir la fecha de corte (hasta el 24 de agosto de 2026)
end_date = '2026-08-24'

# Definir la fecha de inicio (desde el 25 de agosto de 2021)
start_date = '2021-08-25'

# Descargar los datos históricos
df_nbi = yf.download(ticker, start=start_date, end=end_date)

# Mostrar las primeras filas del DataFrame
print(f"\n\nDatos de NBI desde {start_date} hasta {end_date}:")
display(df_nbi.head())
```

### Añadir nuevas columnas al dataset

```python
# 1. Retornos Logarítmicos
df_nbi['Log_Returns'] = np.log(df_nbi['Close'] / df_nbi['Close'].shift(1))

# 2. Volumen Monetario
df_nbi['Monetary_Volume'] = df_nbi['Volume'] * df_nbi['Close']

# Mostrar las primeras filas con las nuevas columnas
print("DataFrame con las nuevas columnas:")
display(df_nbi.head())
```

### Crear una tabla aparte con estadísticas

```python
# Calcular las métricas solicitadas

# Retorno promedio diario
mean_daily_return = df_nbi['Log_Returns'].mean()

# Retorno acumulado
cumulative_return = (df_nbi['Close'].iloc[-1] / df_nbi['Close'].iloc[0]) - 1

# Retorno anualizado (asumiendo 252 días hábiles en un año para retornos logarítmicos)
annualized_return = mean_daily_return * 252

# Desviación estándar de retornos diarios
std_daily_returns = df_nbi['Log_Returns'].std()

# Volatilidad anualizada
annualized_volatility = std_daily_returns * np.sqrt(252)

# Varianza de retornos diarios
variance_daily_returns = df_nbi['Log_Returns'].var()

# Asimetría de retornos diarios
skewness_daily_returns = df_nbi['Log_Returns'].skew()

# Curtosis de retornos diarios
kurtosis_daily_returns = df_nbi['Log_Returns'].kurt()

# Crear un DataFrame para las estadísticas
stats_df = pd.DataFrame({
    'Métrica': [
        'Retorno promedio diario',
        'Retorno acumulado',
        'Retorno anualizado',
        'Desviación estándar de retornos diarios',
        'Volatilidad anualizada',
        'Varianza de retornos diarios',
        'Asimetría de retornos diarios',
        'Curtosis de retornos diarios'
    ],
    'Valor': [
        mean_daily_return,
        cumulative_return,
        annualized_return,
        std_daily_returns,
        annualized_volatility,
        variance_daily_returns,
        skewness_daily_returns,
        kurtosis_daily_returns
    ]
})

# Mostrar la tabla de estadísticas
print("Tabla de Estadísticas de Retornos de NBI:")
display(stats_df)
```

```python
df_nbi.to_csv('df_nbi.csv', index=True)
stats_df.to_csv('nbi_stats.csv', index=False)
print("df_xbi.csv and xbi_stats.csv have been downloaded.")
```
